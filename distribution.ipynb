```python
from string import punctuation
from re import sub
import pandas as pd
```

## Дистрибуция во всему IAVD

```python
iavd = pd.read_csv('all_iavd.csv')
```

```python
# переведем в нижний регистр
iavd['entry'] = iavd['entry'].str.lower()

# удалим случаи, когда отдельно основа и корень
iavd = iavd[~iavd['entry'].str.endswith('-')]
```

### Слова с гортанной смычкой

```python
iavd_words = iavd['entry']
iavd_words_with_gs = iavd_words[iavd_words.str.contains('ʔ')]
```

```python
# два раза гортанная смычка
iavd_words_with_gs.str.contains('ʔ.+ʔ').sum()
```

#### Обработка двойных

Каждый вход с двумя гортанными смычками раздваиваем и одно из ее появлений меняем на GS чтобы правильно считать дистрибуцию

```python
iavd_double_gs = iavd_words_with_gs[iavd_words_with_gs.str.contains('ʔ.+ʔ')]
iavd_words_with_gs_corrected = iavd_words_with_gs[~iavd_words_with_gs.str.contains('ʔ.+ʔ')]
iavd_words_with_gs_corrected = pd.concat((iavd_words_with_gs_corrected, iavd_double_gs.apply(lambda x: sub(r'(ʔ.+)ʔ', r'\1GS', x))))
iavd_words_with_gs_corrected = pd.concat((iavd_words_with_gs_corrected, iavd_double_gs.apply(lambda x: sub(r'ʔ(.+ʔ)', r'GS\1', x))))
```

### Не встречается ни разу с

```python
# создаем набор символов
chars_vocab = set()
for entry in iavd['entry']:
    chars_vocab.update(set(entry))

chars_vocab = chars_vocab - set(punctuation) - set((' ', '˚', 'ь', 'ъ', 'ʼ', '̆')) # убираем явно нерелевантные символы

# правый контекст
print('Правый контекст: ', end='')
for char in chars_vocab:
    if iavd_words_with_gs_corrected.str.contains(f'ʔ{char}').sum() == 0:
        print(char, end=', ')

print()

# левый контекст
print('Левый контекст: ', end='')
for char in chars_vocab:
    if iavd_words_with_gs_corrected.str.contains(f'{char}ʔ').sum() == 0:
        print(char, end=', ')
```
